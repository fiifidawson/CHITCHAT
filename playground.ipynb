```python
import json
from pathlib import Path

# Path to your JSON file (top-level should be a list of objects)
json_path = Path("output/obtained_lit.json")  # <-- change this

with json_path.open("r", encoding="utf-8") as f:
    data = json.load(f)

if not isinstance(data, list):
    raise TypeError(f"Expected a list at the top level, got {type(data).__name__}")

count = len(data)
print(f"{json_path.name} contains {count} JSON objects")
```
