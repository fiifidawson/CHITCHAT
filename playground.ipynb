```python
import json

def find_json_error(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read()
    
    # Try to parse and catch the exact error
    try:
        json.loads(content)
        print("JSON is valid!")
    except json.JSONDecodeError as e:
        print(f"Error at line {e.lineno}, column {e.colno}")
        print(f"Character position: {e.pos}")
        
        # Show context around the error
        start = max(0, e.pos - 500)
        end = min(len(content), e.pos + 500)
        
        print("\nContext around error position:")
        print("..." + content[start:e.pos] + "<<<ERROR HERE>>>" + content[e.pos:end] + "...")
        
        # Show the specific line
        lines = content.split('\n')
        if e.lineno <= len(lines):
            print(f"\nLine {e.lineno}:")
            print(lines[e.lineno-1])

find_json_error('/mloscratch/users/arni/chitchat/CHITCHAT/output/obtained_lit_part_1.json')
```
